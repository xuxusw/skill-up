**Меры объема и центральной тенденции**:  
count - количество заполненных строк (объем выборки).  
mean - среднее арифметическое.  
50% - медиана (центральное значение).  

  
**Меры разброса (вариации)**:  
std - стандартное отклонение (насколько числа в среднем далеки от среднего арифметического).  

  
**Меры положения (квантили и границы)**:  
min / max - минимальное и максимальное значения.  
25% / 75% - первый и третий квартили (границы, отсекающие четверть самых маленьких и четверть самых больших значений).  

Группировка данных:

```python
# средний возраст отдельно для мужчин и для женщин

df.groupby('Sex')['Age'].mean()
```

```python
df.groupby('Sex')['Age'].describe()
```

```python
# сгруппировать и по полу и по тому выжил человек или нет + посчитать среднюю медиану по возрасту

# 1 - передаем список колонок, по которым группируем: ['Sex', 'Survived']
# 2 - что считаем: ['Age']
# 3 - список агрегатных функций, которые хотим посчитать: agg(['mean', 'median']

df.groupby(['Sex', 'Survived'])['Age'].agg(['mean', 'median'])
```

```python
# для дискретных данных: количество элементов данных каждой группы - shortcut того же если сделать groupby и count

df['Sex'].value_counts()
```

```python
df.groupby('Sex')['Sex'].count()
```

```python
# функция корреляции между всеми ЧИСЛОВЫМИ (numeric_only=True) столбцами в датафрейме.

df.corr(numeric_only=True)
```

Изменение данных:

```python
# временный датафрейм (такой же как исходный)

tmp_df = df.copy(deep=True)
```

```python
# 1 способ изменить значение какого-то столбца - присвоить ему константное значение

tmp_df['Pclass'] = 1

tmp_df['Pclass'].value_counts()
```

```python
df['Pclass'].value_counts()
```

```python
# 2 способ - присвоение Series или массива

tmp_df['isAdult'] = tmp_df['Age'] >= 18  # булевая маска

tmp_df.head(10)
```

```python
# изменение значений не всего столбца, а только определенных записей
# (выводим пассажиров которые старше 18 и выжили)

tmp_df['isAdultSurvived'] = tmp_df['isAdult']
tmp_df.loc[tmp_df['Survived'] == 0, 'isAdultSurvived'] = False

tmp_df.head(20)
```

```python
# переименовать название столбца

tmp_df.rename(
    columns={'isAdultSurvived': 'AdultSurvived'}
).head(1)
```

```python
# чтобы не передавать словарь, можно использовать функции

# перевести названия всех столбцов в lowercase
tmp_df.rename(columns=str.lower).head(1)
```

```python
# изменение с помощью lambda функции (apply)

tmp_df['lowercase_name'] = tmp_df['Name'].apply(lambda x: x.lower())

tmp_df['lowercase_name']
```

```python
# для строк есть еще более быстрые решения - векторизированные функции для строк

tmp_df['lowercase_name_2'] = tmp_df['Name'].str.lower()

tmp_df['lowercase_name_2']
```
