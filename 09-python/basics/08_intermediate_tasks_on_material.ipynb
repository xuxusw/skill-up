```python
first_number = float(input())
second_number = float(input())
operation = input() # +, -, /, *, mod, pow, div

match operation:
    case '+':
        print(first_number + second_number)
    case '-':
        print(first_number - second_number)
    case '/' if second_number != 0:
        print(first_number / second_number)
    case '*':
        print(first_number * second_number)
    case 'mod' if second_number != 0:
        print(first_number % second_number)
    case 'pow':
        print(first_number ** second_number)
    case 'div' if second_number != 0:
        print(first_number // second_number)
    case '/' | 'mod' | 'div' if second_number == 0:
        print("Деление на 0!")
    case _:
        print("Неизвестная операция")
```
