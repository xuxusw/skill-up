or, and, not


Часто логические значения возникают в результате сравнения чисел.  

Операции сравнения: <, <=, >, >=, ==, !=

```python
a = int(input())
print(a > 0)
```

```python
a = int(input())
print(a >= 10 and a < 100)
```

```python
x1, x2, x3 = False, True, False
not x1 or x2 and x3
```

```python
True or x2 and x3
```

```python
((not x1) or x2) and x3
```

```python
x = 5
y = 10
y > x * x or y >= 2 * x and x < y
```

```python
a = True
b = False
a and b or not a and not b
```
