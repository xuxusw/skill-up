Условия  
if, else, elif

```python
x = int(input())

if x % 2 == 0:
    print('чет')
else:
    print('нечет')
```

```python
a = int(input())
b = int(input())

if b != 0:
    print(a / b)
elif b < 0: 
    print(a + b)
else:
    print(a - b)
```

```python
A = int(input()) # здоровье
B = int(input()) # пересып
H = int(input()) # аня

if H < A:
    print('Недосып')
elif H > B:
    print('Пересып')
else:
    print('Это нормально')
```

```python
year = int(input())

if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print('Високосный')
else: 
    print('Обычный')
```
