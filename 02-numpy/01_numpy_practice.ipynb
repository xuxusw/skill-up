Индексы и одномерные массивы

```python
arr = np.array([1, -2, 3, -4, 5])
arr[0] = 0
print(arr[2])
print(arr[0:2]) # срез
print(arr[::-1]) # изменить порядок элементов в массиве
print(arr[arr < 2])
print(arr[(arr < 2) & (arr > 0)])
print(arr[(arr > 4) | (arr < 0)])
arr[1:4] = 0 # обнулить элементы
print(arr)
```

При применении определенных функций Numpy, таких как np.mean (),max() и т.д. мы можем указать, по какой оси мы хотим вычислять значения.

Для axis = 0 это означает, что мы применяем функцию к каждому «столбцу» или ко всем значениям, которые встречаются по вертикали.

Для axis = 1 это означает, что мы применяем функцию к каждой «строке» или ко всем значениям по горизонтали.

```python
matrix.sum()
```

```python
matrix
```

```python
print(matrix.max())
print(matrix.max(axis=0)) # наибольший элемент по каждой строке
```

```python
np.mean(matrix, axis=0) # среднее значение идет по строкам (1 + 4 + 7) / 3 = 4
```

```python
np.mean(matrix, axis=1) # среднее значение идет по столбцам (1 + 2 + 3) / 3 = 2
```

Специальные операции

```python
matrix
```

```python
# транспонирование

matrix.T
```

```python
# матрицу в массив

matrix.flatten()
```

```python
# обратная матрица
# ! у вырожденной (singular) матрицы определитель (детерминант) равен нулю, обратной матрицы не существует

np.linalg.inv(matrix)
```

```python
# приближенное решение по методу Мура-Пенроуза (для любых матриц (даже вырожденных и прямоугольных))

np.linalg.pinv(matrix)
```

```python
# след матрицы

np.trace(matrix)
```

```python
# определитель матрицы

np.linalg.det(matrix)
```

```python
# ранг матрицы

np.linalg.matrix_rank(matrix)
```

```python
# собственные числа и векторы матрицы

eigenvalues, eigenvectors = np.linalg.eig(matrix)
print(eigenvalues)
print()
print(eigenvectors)
```
